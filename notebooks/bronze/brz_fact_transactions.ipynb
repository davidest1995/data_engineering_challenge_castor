# Ingesta Raw -> Capa Bronze
Lectura de archivos raw y guardado en formato Parquet sin transformaciones.

```python
input_path = 'file:/Workspace/Users/davidest1995@gmail.com/data_engineering_challenge_castor/data'
```

```python
df_trans = spark.read.csv(f'{input_path}/transacciones_raw.csv', header=True, inferSchema=True)
df_trans = df_trans.withColumn('fecha_ingesta', F.current_date())


df_trans.write.mode('overwrite').partitionBy('fecha_ingesta').saveAsTable('castor.bronze.brz_fact_transactions')
display(df_trans.limit(5))
```
