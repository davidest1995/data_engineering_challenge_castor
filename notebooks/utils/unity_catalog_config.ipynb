# Configuraciones del unity catalog
Configuracion básica de los datos en el unity catalog.

```python
%sql
create catalog if not exists castor;
create schema if not exists castor.bronze;
create schema if not exists castor.silver;
create schema if not exists castor.gold;
```

```python
%sql
CREATE OR REPLACE FUNCTION castor.default.redact_pii_string(p STRING)
    RETURNS STRING
    RETURN CASE
        WHEN is_member('pii_readers') THEN p
        ELSE 'COLUMNA SENSIBLE'
    END;
```

```python
%sql
CREATE OR REPLACE FUNCTION castor.default.edact_pii_identity_string(p STRING)
    RETURNS STRING
    RETURN CASE
        WHEN is_member('pii_readers') THEN p
        ELSE SHA2(p, 256)
    END;
```
