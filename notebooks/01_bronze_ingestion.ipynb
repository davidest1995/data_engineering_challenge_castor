# Ingesta Raw -> Capa Bronze
Lectura de archivos raw y guardado en formato Parquet sin transformaciones.

```
import pyspark.sql.functions as F

base_path = 'file:/Workspace/Repos/your_email/castor-data-engineer/data'

df_trans = spark.read.csv(f'{base_path}/raw/transacciones_raw.csv', header=True, inferSchema=True)
df_trans = df_trans.withColumn('fecha_ingesta', F.current_date())
df_trans.write.mode('overwrite').partitionBy('fecha_ingesta').parquet(f'{base_path}/bronze/transacciones')
```

```
df_user = spark.read.json(f'{base_path}/raw/usuarios_raw.json')
df_user = df_user.withColumn('fecha_ingesta', F.current_date())
df_user.write.mode('overwrite').parquet(f'{base_path}/bronze/usuarios')
```
